## Loli Mart Supermarket 

### Overview
This is designed to provide staff of Loli Mart an automational system that provides the records of every transactions at a particular point in time. This shows a receipt of the list of items customers purchase in addition to the cost of items. 

## Parameters
#### This system has the following key parameters:

* product_id (str): unique identifier for the products
* name (str): Name of the product
* price (float): Actual selling price of a product
* product (str): Name of the product
* quantity (int): Number of products purchased
* order_id (str): unique identifier for the orders
* sales_person (str): Cashier
* order_date_time (Datetime): The particular date and time a product was purchased in yyyy/mm/dd format

```python
import datetime

class Product:
    """
     This brings out a product with their product_id, Name, and price

        Attributes
        
        product_id (str): unique identifier for the products
        name (str): Name of the product
        price (float): Actual selling price of a product

    """
    def __init__(self, product_id, name, price):
        self.product_id = product_id
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} (ID: {self.product_id}), Price: ₵{self.price:.2f}"

class OrderItem:
    """
    This brings out an order item with their product and quantity

        Attributes
        
        product (str): Name of the product
        quantity (int): Number of products purchased

    """
    def __init__(self, product, quantity):
        self.product = product
        self.quantity = quantity
        self.subtotal = product.price * quantity

    def __str__(self):
        return f"{self.quantity} x {self.product.name} - ₵{self.subtotal:.2f}"

class Order:
    """
     This brings out an order with their order_id, salesperson, order_date_time

        Attributes

        order_id (str): unique identifier for the orders
        customer_id (str): unique identifier for the customers
        sales_person (str): Cashier
        order_date_time (Date): The particular date and time a product was purchased in  yyyy/mm/dd format
        
    """
    def __init__(self, order_id, salesperson, order_date_time=None):
        self.order_id = order_id
        self.salesperson = salesperson
        self.order_date_time = datetime.datetime.now()
        self.items = []
        self.total_sales = 0

    def add_item(self, product, quantity):
        item = OrderItem(product, quantity)
        self.items.append(item)
        self.total_sales += item.subtotal

    def display_order(self):
        print(f"Order ID: {self.order_id}")
        print(f"Salesperson: {self.salesperson}")
        print(f"Date/Time: {self.order_date_time}")
        print("Items:")
        for item in self.items:
            print(f"  {item}")
        print(f"Total Sales: ₵{self.total_sales:.2f}")

class Supermarket:
    """
     This brings out a name of the supermarket

    Attributes
    name (str): The name of the supermarket
        
    """
    def __init__(self, name):
        self.name = name
        self.products = {}
        self.orders = {}
        self.order_counter = 1

    def add_product(self, product):
        self.products[product.product_id] = product
        print(f"{product.name} added to {self.name}'s inventory.")

    def create_order(self, salesperson):
        order = Order(self.order_counter, salesperson)
        self.orders[self.order_counter] = order
        self.order_counter += 1
        return order

    def find_product(self, product_id):
        return self.products.get(product_id)

    def process_order(self, order):
        order.display_order()
```

```python
##naming our class as LoliSupermarket
supermarket = Supermarket("Loli Mart")
```

```python
## product details
product1 = Product(1, "Milk", 8.00)
product2 = Product(2, "Bread", 20.00)

supermarket.add_product(product1)
supermarket.add_product(product2)

print("\n")

#customer 1 bought two items (Milk and Bread)
order1 = supermarket.create_order("Eric")
order1.add_item(product1, 2)
order1.add_item(product2, 1)

supermarket.process_order(order1)
```
